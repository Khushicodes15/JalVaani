## JalVaani AI — Day 3 Research Finding

### Contamination Risk Classification Results

**Data source:** CGWB, Rajya Sabha Session 267, AU-1211 (official government data)

| Contaminant | ROC-AUC | Interpretation |
|---|---|---|
| Fluoride | 0.80 | Genuine generalization — geologically driven |
| Arsenic | 0.41 | Below chance — insufficient positive states |
| Nitrate | 0.50 | Random — agricultural features absent |

*(ROC-AUC from leave-states-out evaluation: entire states held out, state-identifying features excluded. Random-split metrics are 1.000 by construction — labels are state-constant — and are reported only as a data-integration check.)*

### Why the asymmetry matters

**Fluoride (AUC 0.80):** Fluoride contamination is geologically
determined — arid hard-rock aquifers, deep groundwater, specific
lithology. Well-level features (depth, location, seasonality) carry
real signal that generalizes to unseen states. This is a defensible,
meaningful result from state-aggregate labels.

**Arsenic (AUC 0.41):** With only 3 positive states (Bihar, UP,
West Bengal), leave-states-out cross-validation has almost no
positive examples in training when those states are held out.
The model cannot learn the Indo-Gangetic Plain pattern from
the remaining states alone. This is a data structure problem,
not a modeling problem.

**Nitrate (AUC 0.50):** Nitrate contamination is driven by
agricultural fertilizer intensity — a feature not present in
groundwater level records. No amount of modeling can recover
signal that was never collected.

### Key insight

State-aggregate labels support cross-state generalization only
for geologically-determined contaminants with sufficient positive
states. Arsenic and nitrate require well-level quality measurements
(CGWB/India-WRIS individual sample data) to be modeled meaningfully.

**This finding directly motivates JalVaani's next research direction:
integrating well-level water quality measurements from India-WRIS
to replace state-aggregate labels.**
